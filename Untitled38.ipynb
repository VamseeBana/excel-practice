```python
import statistics
import matplotlib.pyplot as plt

member_name=["vamc","ravi","nani","ram","rao"]
age=[22,25,24,23,26]
workout_hours=[2,1.5,2.5,1,2]
caloris_burned=[630,450,300,270,400]
print("fitness dataset  ")

for i in range(len(member_name)):
  print(member_name[i],
        age[i],
           workout_hours[i],
        caloris_burned[i])

average_workout=sum(workout_hours)/len(workout_hours)
print("average_workout",average_workout)

max_caloris=max(caloris_burned)
print("max_caloris",max_caloris)

min_caloris=min(caloris_burned)
print("min_caloris",min_caloris)

median_caloris=statistics.median(caloris_burned)
print("median_caloris",median_caloris)

std_dev=statistics.stdev(caloris_burned)
print("std_dev",std_dev)

var=statistics.variance(caloris_burned)
print("var",var)

index=caloris_burned.index(max_caloris)
print(member_name[index])
index=caloris_burned.index(min_caloris)
print(member_name[index])

plt.bar(member_name,caloris_burned)
plt.xlabel("member_name")
plt.ylabel("caloris_burned")
plt.title("fitness analysis")
plt.show()

plt.scatter(workout_hours,caloris_burned)
plt.xlabel("workout_hours")
plt.ylabel("caloris_burned")
plt.title("fitness analysis")
plt.show()

plt.boxplot(caloris_burned)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statistics
df=pd.read_csv("/content/cleangym.csv")
df.head()
df.info()
df.shape
```

```python
avg_rate=df["fitness_participation_rate"].mean()
print("=",avg_rate)
med_rate=df["fitness_participation_rate"].median()
print("=",med_rate)
```

```python
df_groupby_region = df.groupby("region")["number_of_gyms"].sum()
print(df_groupby_region)
df_groupby_region.plot(kind="pie")
plt.show()
```

```python
df_groupby_country_max = df.groupby("country")["number_of_gyms"].sum().max()
print(df_groupby_country_max)

print(df_groupby_country_max)
```

```python
df_groupby_country = df.groupby("country")["number_of_gyms"].sum()
print(df_groupby_country)
df_groupby_country.plot(kind="bar")
plt.show()
```
